```python
def calculate_lb_agar(target_volume_ml, language="en"):
    """
    LB Agar hazırlama protokolü ve malzeme hesaplayıcı.
    LB Agar preparation protocol and ingredient calculator.
    """
    ratio = target_volume_ml / 1000.0
    lb_broth_powder = 25.0 * ratio
    agar_powder = 15.0 * ratio
    estimated_plates = int(target_volume_ml // 20)  # ~20 mL per plate

    if language.lower() == "tr":
        print(f"\n--- {target_volume_ml} mL LB Agar Hazırlama Protokolü ---")
        print(f"1. Distile Su: {target_volume_ml} mL")
        print(f"2. Tartılacak LB Broth tozu: {lb_broth_powder:.2f} g")
        print(f"3. Tartılacak Agar-agar: {agar_powder:.2f} g")
        print(f"-> Tahmini Verim: Yaklaşık {estimated_plates} adet standart petri kabı.")
        print("\nProtokol:")
        print("- Tozları distile suya ekleyip manyetik karıştırıcıda çözdürün.")
        print("- 121°C'de 15 dakika otoklavlayın.")
        print("- 50°C'ye kadar soğutup steril kabin içinde steril petrilere paylaştırın.")
    else:
        print(f"\n--- Protocol for {target_volume_ml} mL LB Agar ---")
        print(f"1. Distilled Water: {target_volume_ml} mL")
        print(f"2. LB Broth Powder: {lb_broth_powder:.2f} g")
        print(f"3. Agar-agar Powder: {agar_powder:.2f} g")
        print(f"-> Yield: Approximately {estimated_plates} standard Petri plates.")
        print("\nProcedure:")
        print("- Dissolve powders in distilled water using a magnetic stirrer.")
        print("- Autoclave at 121°C for 15 minutes.")
        print("- Cool to ~50°C and pour into sterile Petri dishes inside a laminar flow hood.")

# Kullanıcı girişleri / User prompts:
lang_choice = input("Dil seçin / Select language (TR / EN): ").strip()
vol_choice = float(input("Hacim girin / Enter volume (mL): "))

calculate_lb_agar(vol_choice, language=lang_choice)
```
